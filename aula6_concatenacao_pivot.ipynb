```python
import pandas as pd 
import numpy as np
import os
pd.options.display.float_format = '{:,.2f}'.format
```

```python
vendas_db_caminho = 'db\\Base_analitica_vendas_periodo.xlsx' 
```

```python
vendas_db_dezembro = pd.read_excel(vendas_db_caminho, sheet_name='Dez')
vendas_db_janeiro = pd.read_excel(vendas_db_caminho, sheet_name='Jan')
vendas_db_fevereiro = pd.read_excel(vendas_db_caminho, sheet_name='Fev')
```

```python
vendas_db_dezembro = vendas_db_dezembro.rename(columns={
    "DATA": "DATAEMISSAO"
})
```

```python
vendas_db = pd.concat(
    [
        vendas_db_dezembro,
        vendas_db_janeiro,
        vendas_db_fevereiro    
    ],
    ignore_index=True
)
```

```python
vendas_db["MÊS"] = vendas_db.DATAEMISSAO.dt.month_name(locale="pt_BR")
```

```python
vendas_db
```

### Exportação

```python
os.makedirs("exportacoes", exist_ok=True)
```

```python
vendas_db.to_excel("exportacoes\\Base vendas Dez, Jan, Fev.xlsx", index=False)
```

```python
vendas_db.to_csv("exportacoes\\Base vendas Dez, Jan, Fev.csv", sep=";", index=False, encoding="utf-8")
```

```python
vendas_db.to_json("exportacoes\\Base vendas Dez, Jan, Fev.json", orient="records", force_ascii=False, date_format="iso", indent=4)
```
