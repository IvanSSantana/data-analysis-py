```python
import pandas as pd 
import numpy as np
```

```python
vendas_db_caminho = 'db\\Base_analitica_vendas.xlsx' 
```

```python
vendas_db = pd.read_excel(vendas_db_caminho, sheet_name='Base')
```

### Clientes que mais faturaram

```python
gastos_clientes = vendas_db.groupby(["DESCRICAOCLIENTE"])["VALORLIQUIDO"].sum().reset_index()
gastos_clientes = gastos_clientes.sort_values("VALORLIQUIDO", ascending=False)

gastos_clientes.head(10)
```

###  Produtos que mais faturaram por cliente

```python
gastos_clientes = vendas_db.groupby(["DESCRICAOCLIENTE", "NOMEPRODUTO"])[["VALORLIQUIDO", "VOLUME"]].sum().reset_index()
gastos_clientes = gastos_clientes.sort_values("VALORLIQUIDO", ascending=False)

gastos_clientes.head(15)
```

### Agregação de múltiplas métricas por diferentes colunas

```python
clientes = (
    vendas_db.groupby("DESCRICAOCLIENTE")
    .agg({
        "VALORLIQUIDO": "sum",
        "VOLUME": "sum",
    })
    .reset_index()
)

clientes.head(5)
```

### Agregação de múltiplas métricas por coluna

```python
vendas_db.groupby(["DESCRICAOGRUPOPRODUTO"])["VALORLIQUIDO"].agg(["sum", "mean", "min", "max"]).reset_index().sample(7)
```
