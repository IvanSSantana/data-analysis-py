```python
import pandas as pd 
```

```python
vendas_db_caminho = 'db\\Base venda bruta marco.csv' 
```

```python
vendas_db = pd.read_csv(vendas_db_caminho, sep=";")
```

```python
vendas_db_tratada = vendas_db.copy()
```

```python
vendas_db_tratada["VOLUMESELLOUT"] = vendas_db_tratada["VOLUMESELLOUT"].fillna(0) 
vendas_db_tratada["UF"] = vendas_db_tratada["UF"].fillna("SEM REGIÃO") 
vendas_db_tratada = vendas_db_tratada.dropna(subset=["DATA"])
vendas_db_tratada = vendas_db_tratada.drop_duplicates()
```

```python
vendas_db_tratada.isna().sum()
```

```python
vendas_db_tratada.duplicated().sum()
```

```python
vendas_db_tratada["UF"] = vendas_db_tratada["UF"].str.upper().str.strip()       
vendas_db_tratada["DATA"] = pd.to_datetime(vendas_db_tratada["DATA"])
vendas_db_tratada = vendas_db_tratada.drop(columns=["DATACARGA"])
```
