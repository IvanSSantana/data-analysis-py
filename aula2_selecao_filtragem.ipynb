# Filtragem

```python
vendas_db[vendas_db["VALORLIQUIDO"] >= 10_000].drop_duplicates()
```

```python
vendas_db[vendas_db["DESCRICAOTIPOFATURAMENTO"].isin(["DEVOLUÇÃO DE VENDA", "DEVOLUÇÃO DE COMPRA"])].sample(10)
```

```python
vendas_db[
    (vendas_db["DESCRICAOTIPOFATURAMENTO"] == "DEVOLUÇÃO DE VENDA") |
    (vendas_db["DESCRICAOTIPOFATURAMENTO"] == "DEVOLUÇÃO DE COMPRA") &
    (vendas_db["VOLUME"] <= -50)
    ].sample(8)
```

```python
vendas_db.query("VOLUME >= 1000 & VOLUME < 2000").head(5)
```

```python
tipo_faturamento = 1
volume = 1000

vendas_db.query(f"TIPOFATURAMENTO == {tipo_faturamento} | VOLUME >= @volume").head(5)
```

O uso dos dois tipos de interpolação é contraindicado, use o de sua preferência. Entretanto o mais recomendado para segurança é com @
