### Numpy.Where x Pandas.Apply

O where() proporciona melhor segurança e perfomance do que o apply(), no geral.

```python
vendas_db["VALORUNITARIO"] = np.where(
    vendas_db["VOLUME"] > 0, 
    vendas_db["VALORLIQUIDO"] / vendas_db["VOLUME"],
    0 
)
```

```python
vendas_db[["VALORLIQUIDO", "VOLUME", "VALORUNITARIO"]].sample(5)
```

```python
vendas_db["CLASSIFICACAO_VALOR"] = np.select(
    [
        vendas_db["VALORLIQUIDO"] >= 5000,
        vendas_db["VALORLIQUIDO"] >= 3500
    ],
    [
        "Alto",
        "Mediano"
    ],
    default="Baixo"
)
```
